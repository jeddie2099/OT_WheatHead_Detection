# YOLO models for wheat head detection in GWHD 2021

This notebook contains the training and validation of the YOLO models used for wheat head detection, both with the original dataset and with its version modified by the domain adaptation techniques. The YOLO models were loaded with the Ultralytics library and pre-trained models were used to allow for a fast convergence on the training with the GWHD 2021 dataset. Different [pre-trained models](https://docs.ultralytics.com/models/) can be downloaded directly from Ultralytics, with different YOLO versions and model sizes available. 

## YOLO model with the original dataset

### Model training 

```python
import torch
from ultralytics import YOLO
```

```python
# Load any pretrained model you have downloaded from the Ultralytics site, so we can perform fine-tuning with gwhd 2021 dataset
model_path = '../gwhd_2021/yolo12s.pt'
data_path = '../gwhd_2021/gwhd_2021.yaml'

# Define the folder where the fine-tuned model will be saved at the end of the training 
save_path =  '../gwhd_2021/yolov12/trainings'

# Load a COCO pretrained YOLO model
model = YOLO(model_path) 

# Train the model on the gwhd 2021 dataset and modify the parameters as needed 
results = model.train(data=data_path, epochs=100, imgsz=640, patience=10, seed=0, project=save_path)
```

### Validation of the model

```python
# Path to the model weights to perform validation
best_model = '../gwhd_2021/yolov12/trainings/train/weights/best.pt'
data_path = '../gwhd_2021/gwhd_2021.yaml'

# Define a folder to save the results of the test metrics
save_path =  '../gwhd_2021/yolov12/tests'

# Load the weights for the best performing epoch in the training above
model = YOLO(best_model)

# Obtain results for the test set
validation_results = model.val(data=data_path, imgsz=640, split="test", project=save_path)
```

## YOLO with the OT modified dataset

### Model Training

```python
import torch
from ultralytics import YOLO
```

```python
# Load any pretrained model you have downloaded from the Ultralytics site, so we can perform fine-tuning with gwhd 2021 dataset
model_path = '../gwhd_2021/yolo12s.pt'
data_path = '../gwhd_2021/gwhd_2021_OT.yaml'

# Define the folder where the fine-tuned model will be saved at the end of the training 
save_path =  '../gwhd_2021/yolov12/trainings'

# Load a COCO pretrained YOLO model
model = YOLO(model_path) 

# Train the model on the version of gwhd 2021 dataset modified by OT domain adaptation and modify the parameters as needed 
results = model.train(data=data_path, epochs=100, imgsz=640, patience=10, seed=0, project=save_path)
```

### Validation of the model

```python
# Path to the model weights to perform validation 
best_model = '../gwhd_2021/yolov12/trainings/train2/weights/best.pt'
data_path = '../gwhd_2021/gwhd_2021_OT.yaml'

# Define a folder to save the results of the test metrics 
save_path =  '../gwhd_2021/yolov12/tests'

# Load the weights for the best performing epoch in the training above
model = YOLO(best_model)

# Obtain results for the test set
validation_results = model.val(data=data_path, imgsz=640, split="test", project=save_path)
```
