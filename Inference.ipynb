# Inference for wheat head detection with trained YOLO model

## Inference for unmodified image

```python
from utils import wheat_detection
import matplotlib.pyplot as plt
import cv2

trained_model = '../gwhd_2021/yolov12/trainings/train/weights/best.pt'
img = "../gwhd_2021/Original/test/images/e4a42832b0f767fdfb3193cffe4a86538e3040b9789a9bd053f93d434dfa5201.png"
inference_path = "figs/inference_boxes.png"
annotation_path = "../gwhd_2021/Original/test/labels/e4a42832b0f767fdfb3193cffe4a86538e3040b9789a9bd053f93d434dfa5201.txt"
output_path =  "figs/predicted_and_gtruth.png"

inference = wheat_detection(trained_model, img, inference_path, annotation_path, output_path)
plt.imshow(cv2.cvtColor(inference, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
